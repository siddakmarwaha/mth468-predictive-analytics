### 1. Write a procedure that finds the sum of the first n integers.

 **a. Write this as an iterative procedure.**

```python
def iter_sum(n):
    amt = 0
    for i in range(n+1):
        amt = amt+i
    return amt
iter_sum(10)
```

**b. Write this as a recursive procedure.**

```python
def rec_sum(n):
    if n == 0:
        return 0
    amt = rec_sum(n-1) + n
    return amt
rec_sum(10)
```

### 2. Estimate the value of 𝜋 with an accuracy of 0.0001

```python
def taylor_pi(accuracy):
    pi = 0
    n = 0
    while True:
        term = (-1)**n / (2 * n + 1)
        pi += term
        if abs(4 * pi - math.pi) < accuracy:
            break
        n += 1
    return 4 * pi, n

estimated_pi, n = taylor_pi(0.0001)
print("Estimated value of π:", estimated_pi, 'with', n, 'terms')
```

### 3. Write a procedure that produces Pascal’s Triangle.

```python
import math
import numpy as np
```

```python
def pascal_tri(n):
    for i in range(n):
        print()
        for j in range(i+1):
            print(math.comb(i,j), end = ' ')
pascal_tri(8)
```

### 4. Write a procedure that finds the prime factors of a number.

```python
def prime_factors(n):
    i = 2
    factors = []
    while i * i <= n: # If n is not divisible by any number less than or equal to the square root of n, then it cannot have any factors other than itself and 1, and thus it must be prime.
        if n % i != 0:
            i += 1
        else:
            n //= i
            factors.append(i)
    if n > 1: # Include the number that is left
        factors.append(n)
    return factors
```
