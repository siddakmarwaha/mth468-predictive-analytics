### b) 1. 1.

```python
# Define valid values of k
k_values = np.arange(1, 51)

num_folds = [2, 5, 10, 15, 25, 50]

mean_accuracy_scores = []

# Iterate over each value of k
for k in k_values:
    k_accuracy_scores = []
    
    # Iterate over each number of folds
    for folds in num_folds:
        knn = KNeighborsClassifier(n_neighbors=k)
        accuracy_scores = cross_val_score(knn, iris.data[:, :2], iris.target, cv=folds, scoring='accuracy')
        mean_accuracy = np.mean(accuracy_scores)
        k_accuracy_scores.append(mean_accuracy)
    
    # Append list of mean accuracy scores for current k to the main list
    mean_accuracy_scores.append(k_accuracy_scores)

# Convert to numpy array for plotting
mean_accuracy_scores = np.array(mean_accuracy_scores)

# Plot error curve
plt.figure(figsize=(10, 6))
for i, folds in enumerate(num_folds):
    plt.plot(k_values, mean_accuracy_scores[:, i], marker='o', linestyle='-', label=f'{folds}-fold Cross Validation')
plt.xlabel('Number of Neighbors')
plt.ylabel('Mean Accuracy')
plt.title('Error Curve of kNN')
plt.legend()
plt.grid()
plt.show()
```

```python
X = iris.data[:, :2]  # Considering only the first two features for visualization
y = iris.target

# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Plot dataset
plt.figure()
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis')
plt.xlabel('Sepal length')
plt.ylabel('Sepal width')
plt.title('Iris Dataset')
plt.show()

# k-NN with all distance metrics
k = 5
distance_metrics = ['euclidean', 'manhattan', 'chebyshev']

for metric in distance_metrics:
    knn = KNeighborsClassifier(n_neighbors=k, metric=metric)
    knn.fit(X_train, y_train)

    # Define decision boundary
    h = 0.02  # step size
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
    Z = knn.predict(np.c_[xx.ravel(), yy.ravel()])

    # Plot result
    Z = Z.reshape(xx.shape)
    plt.figure()
    plt.contourf(xx, yy, Z, cmap='viridis', alpha=0.4)

    # Plot training points
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis')
    plt.xlabel('Sepal length')
    plt.ylabel('Sepal width')
    plt.title(f'Decision Boundary: {metric}')
    plt.show()
```

```python
inertia = []

# Calculate inertia
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=1)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

# Plot error curve
plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Error Curve')
plt.show()
```

### a)

```python
target = pd.read_excel('target.xlsx')

# Define new data points
new_data = pd.DataFrame({
    'Abscissa': [-2, 1],
    'Ordinate': [-1, -3],
    'Bullseye': ['Yes', 'Yes']
})

# Concatenate data
target = pd.concat([target, new_data], ignore_index=True) # ignore_index appends while continuing the sequence of indices
target
```

### b) 1. and 2.

```python
# Extract features and target variable
X = target[['Abscissa', 'Ordinate']].values
y = target['Bullseye'].map({'Yes': 1, 'No': 0}).values

# Standardizing
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Define SVM kernels
classifiers = {
    'Linear': SVC(kernel='linear'),
    'RBF': SVC(kernel='rbf'),
}

# Plot data and decision boundaries
for kernel, classifier in classifiers.items():
    
    # Train classifier
    classifier.fit(X_scaled, y)
    
    # Plot decision boundary
    plt.figure(figsize=(8, 6))
    x_min, x_max = X_scaled[:, 0].min() - 1, X_scaled[:, 0].max() + 1
    y_min, y_max = X_scaled[:, 1].min() - 1, X_scaled[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1), np.arange(y_min, y_max, 0.1))
    Z = classifier.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=0.4)
    
    # Plot data points
    plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=y, edgecolor='k')
    plt.title(f'SVM with {kernel} Kernel')
    plt.xlabel('Abscissa')
    plt.ylabel('Ordinate')
    plt.show()
```

```python
# Define degrees
degrees = range(1, 7)
max_iter = 10000

for degree in degrees:
    classifier = SVC(kernel='poly', degree=degree, max_iter=max_iter)
    classifier.fit(X_scaled, y)
    
    # Plot decision boundary
    plt.figure(figsize=(8, 6))
    x_min, x_max = X_scaled[:, 0].min() - 1, X_scaled[:, 0].max() + 1
    y_min, y_max = X_scaled[:, 1].min() - 1, X_scaled[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1), np.arange(y_min, y_max, 0.1))
    Z = classifier.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=0.4)
    
    # Plot data points
    plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=y, edgecolor='k')
    plt.title(f'SVM with Polynomial Kernel (degree={degree})')
    plt.xlabel('Abscissa')
    plt.ylabel('Ordinate')
    plt.show()

# 1. Using degree = 2 and C = 0.05, call the support vectors of the model
classifier = SVC(kernel='poly', degree=2, C=0.05)
classifier.fit(X_scaled, y)
support_vectors = classifier.support_vectors_

print("Support Vectors:")
print(support_vectors)

# 2. Using degree = 2 and C = 0.05, call the dual coefficients of the model
dual_coef = classifier.dual_coef_

print("Dual Coefficients:")
print(dual_coef)
```

- From observing the plots, I realized that the SVM classifier with a polynomial kernel of degree 2 showed the most effective performance. Models with odd polynomial degrees (1, 3, and 5) failed to effectively classify the data. In contrast, models with even degrees, particularly as the degree increased from 2 to 4 and then 6, began to include more non-bullseye points within the bullseye classification. So, increasing the polynomial degree led to a more complex decision boundary, resulting in a broader inclusion of points from both classes.

- Each dual coefficient corresponds to a support vector and determines its influence on the decision boundary.


- In the specific case with C = 0.05, the dual coefficients indicate the contribution of each support vector to the decision function, where a negative coefficient suggests a misclassified support vector, and a positive coefficient indicates a correctly classified support vector.

### 5.

- Linear Kernel: This assumes that the data is linearly separable. If the classes are well-separated by a straight line, the linear kernel performs well. However, since the classes are not linearly separable, the linear kernel does poorly to classify the data.


- RBF Kernel: This also captures non-linear decision boundaries. It worked well since the classes are not linearly separable and there were more complex relationships between the features and the target. The RBF kernel performed the best in this case because it adapted to the complex patterns in the data.


- Polynomial Kernel: This can also capture non-linear decision boundaries, but its performance depends on the degree of the polynomial. A lower degree underfits the data, while higher degrees may overfit. In this case, the polynomial kernel with degree 2 performed well because it found a suitable balance between capturing non-linear relationships and avoiding overfitting.


- Sigmoid Kernel: This can be useful when the data has a sigmoidal shape. However, it is sensitive to scaling and may not perform well if the data is not scaled properly. In this case, the sigmoid kernel performed poorly because the data did not have a sigmoidal shape.

```python
# Elbow method to determine the optimal number of clusters (k)
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=1)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

# Plot the error curve (inertia) to find the elbow
plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Error Curve')
plt.show()

# Based on the elbow plot, select the optimal number of clusters (k)
# In this case, we'll choose k=3

# Fit k-means with the optimal number of clusters
kmeans = KMeans(n_clusters=3, random_state=1)
kmeans.fit(X)

# Define meshgrid for decision regions
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1), np.arange(y_min, y_max, 0.1))

# Predict labels for each point in the meshgrid
Z = kmeans.predict(np.c_[xx.ravel(), yy.ravel()])

# Reshape the predictions into the meshgrid shape
Z = Z.reshape(xx.shape)

# Plot decision boundaries and cluster centers
plt.figure(figsize=(10, 8))
plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='viridis', edgecolor='black')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', marker='x', s=100)
plt.xlabel('Abscissa')
plt.ylabel('Ordinate')
plt.title('K-means Clustering with k=3')
plt.show()
```

- SVM with the RBF kernel performed the best amongst all for this dataset. It provided the most accurate classification with zero misclassified points. The RBF kernel is particularly effective when the decision boundary is non-linear, as it flexibly adjusts to the shape of the data.


- While k-NN is a simple algorithm, it didn't perform as well as SVM with the RBF kernel for this dataset. Regardless of the distance metric chosen, k-NN performed poorly for this dataset.


- K-Means clustering showed really good results in grouping similar data points together. In this dataset, k-Means could help identify underlying patterns or groupings in the data pretty well.
